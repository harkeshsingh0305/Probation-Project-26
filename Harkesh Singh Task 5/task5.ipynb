```python
Q1 = df['num of ratings'].quantile(0.25)
Q3 = df['num of ratings'].quantile(0.75)

IQR = Q3-Q1

lower = Q1-(1.5*IQR)
upper = Q3+(1.5*IQR)

outliers = df[(df['num of ratings']<lower) | (df['num of ratings']>upper)]

outliers.head(5)
```

```python
import re



# Function to clean restaurant names
def clean_restaurant_name(name):

    if pd.isna(name):
        return name

    name = str(name)

    # 1. Fix encoding / mojibake
    for _ in range(3):
        try:
            fixed = name.encode("latin1").decode("utf-8")
            if fixed == name:
                break
            name = fixed
        except (UnicodeEncodeError, UnicodeDecodeError):
            break

    # 2. Remove leading/trailing spaces
    name = name.strip()

    # 3. Remove unwanted symbols from beginning
    # Example: #FeelTheROLL -> FeelTheROLL
    # '@ Biryani Central -> Biryani Central
    name = re.sub(r"^[#@'\"]+\s*", "", name)

    # 4. Remove unnecessary special characters from beginning/end
    name = re.sub(r"^[^A-Za-z0-9]+", "", name)
    name = re.sub(r"[^A-Za-z0-9À-ÿ&'().,+\- ]+$", "", name)

    # 5. Replace multiple spaces with one
    name = re.sub(r"\s+", " ", name)

    # 6. Remove spaces before punctuation
    name = re.sub(r"\s+([,.)])", r"\1", name)

    # 7. Remove spaces after opening brackets
    name = re.sub(r"([\(])\s+", r"\1", name)

    return name.strip()


# Apply cleaning
df["restaurant name"] = df["restaurant name"].apply(clean_restaurant_name)

# Check results
df.head(20)
```
